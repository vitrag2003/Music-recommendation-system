```python
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler, MinMaxScaler

sub_features = features.copy()
sub_features.remove('year')

scaler = StandardScaler(with_mean=True, with_std=True)
X_scaled = scaler.fit_transform(X[ sub_features ])

X_scaled = pd.DataFrame(X_scaled, columns=sub_features)
# X_scaled = X_scaled.join(data['year'])

X_scaled.head()
```

```python

pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

plt.figure(figsize=(8, 6))
plt.scatter(X_pca[:, 0], X_pca[:, 1], c=data['year'])
plt.xlabel('First Principal Component')
plt.ylabel('Second Principal Component')
plt.title('PCA')
plt.show()
```

```python
pca.explained_variance_ratio_
```

```python
pca = PCA().fit(X_scaled)

# Calculate explained variance
explained_variance = np.cumsum(pca.explained_variance_ratio_)

# Plot the explained variance
plt.figure(figsize=(8, 5))
plt.plot(range(1, len(explained_variance) + 1), explained_variance, marker='o', linestyle='--')
plt.xlabel('Number of Components')
plt.ylabel('Cumulative Explained Variance')
plt.title('Explained Variance vs Number of Components')
plt.grid()
plt.show()

# Finding optimal number of components that explain desired variance
desired_variance = 0.95  # Adjust this as needed
optimal_n_components = np.argmax(explained_variance >= desired_variance) + 1

print(f"Optimal number of components for {desired_variance*100}% variance: {optimal_n_components}")
```

```python
pca = PCA(n_components=optimal_n_components)
X_pca = pca.fit_transform(X_scaled)

X_pca = pd.DataFrame(X_pca, columns=[f'PC{i}' for i in range(1, optimal_n_components + 1)])
X_pca['year'] = data['year'].values

X_pca.head()
```

```python
from sklearn.pipeline import Pipeline
from sklearn.neighbors import NearestNeighbors

# Create a pipeline
pipeline = Pipeline([
    ('scaler', StandardScaler()),
    ('pca', PCA(n_components=optimal_n_components)),
    # ('knn', NearestNeighbors(n_neighbors=5))
])

pipeline.fit(X[ sub_features ].values)

neigh = NearestNeighbors(n_neighbors=4)
neigh.fit(X_pca)
```

```python
def recommend_songs_by_id(song_id):
    # Get the song's features
    song = X.loc[song_id, sub_features].values.reshape(1, -1)
    # Transform the features using the pipeline
    song_transformed = pipeline.transform(song)
    # Find the nearest neighbors
    
    # song_transformed= song_transformed.reshape(1, -1)
    
    song_transformed = pd.DataFrame(song_transformed, columns=[f'PC{i}' for i in range(1, optimal_n_components + 1)])
    song_transformed['year'] = data.loc[song_id, 'year']    

    distances, indices = neigh.kneighbors(song_transformed)
    # Get the indices of the recommended songs
    recommended_songs = data.iloc[indices[0]].index
    return recommended_songs
# Get recommendations for a song
song_id = '7xPhfUan2yNtyFG0cUWkt8'
recommended_songs = recommend_songs_by_id(song_id)
recommended_songs
```

```python
from sklearn.metrics.pairwise import cosine_similarity

def recommend_songs_by_artist(song_id):
    artist_name = data.loc[song_id, 'artists_name']
    songs_by_artist = data[(data['artists_name'] == artist_name) & (data.index != song_id)].index
    
    # Get the song's features
    cos_sim = []
    for song in songs_by_artist:
        song_features = X.loc[song, sub_features].values.reshape(1, -1)
        song_features_transformed = pipeline.transform(song_features)
        cos_sim.append(cosine_similarity(song_features_transformed, pipeline.transform(X.loc[song_id, sub_features].values.reshape(1, -1))))
        
    cos_sim = np.array(cos_sim).reshape(-1)
    indices = np.argsort(cos_sim)[::-1]
    recommended_songs = songs_by_artist[indices]
    return recommended_songs
```
